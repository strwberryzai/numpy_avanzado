```python
import numpy as np
```

```python
# ejercicio 1

a = np.array([1, 2, 3, 4])
ningun_cero = np.all(a != 0)
print(ningun_cero)  # True
```

```python
# ejercicio 2
a = np.array([0, 0, 2, 0])
alguno_no_cero = np.any(a != 0)
print(alguno_no_cero)  # True
```

```python
# ejercicio 3
a = np.array([1, 2, np.nan, np.inf])
tiene_inf = np.any(np.isinf(a))
tiene_nan = np.any(np.isnan(a))
print(tiene_inf, tiene_nan) 
```

```python
# ejercicio 4
a = np.array([1, 2, np.inf])
tiene_inf = np.any(np.isinf(a))
print(tiene_inf)  # 
```

```python
# ejercicio 5
a = np.array([1, np.nan, 3])
tiene_nan = np.any(np.isnan(a))
print(tiene_nan)  # True
```

```python
# ejercicio 6
a = np.array([1, 3, 5])
b = np.array([2, 3, 4])

print("a >  b:", a > b)
print("a >= b:", a >= b)
print("a <  b:", a < b)
print("a <= b:", a <= b)
```

```python
#Ejercicio 7
a = np.array([1, 7, 13, 105])
print("Bytes totales:", a.nbytes)
```
